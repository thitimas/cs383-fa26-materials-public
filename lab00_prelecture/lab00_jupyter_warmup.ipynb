## 1. What is a Jupyter notebook?

This webpage is called a Jupyter notebook. A notebook is a place to write code and view its results, and
also to write text — like this paragraph.

### 1.1 Text cells

In a notebook, each rectangle containing text or code is called a **cell**.

Text cells (like this one) can be edited by double-clicking on them. They're written in a simple format
called [Markdown](http://daringfireball.net/projects/markdown/syntax) to add formatting and section
headings. You don't need to learn Markdown right now, but you'll pick some up naturally over the semester.

After you edit a text cell, click the "run cell" button at the top that looks like ▶, or hold down
`Shift` + `Return`, to confirm the change. (Try not to delete these instructions while you're practicing!)

### 1.4 Errors

Python is a language, and like natural human languages it has rules. It differs from natural language in
two important ways:

1. The rules are *simple* — you can learn most of them in a few weeks.
2. The rules are *rigid* — a computer running Python won't gloss over small mistakes the way a person
   listening to you speak would.

Whenever you write code, you'll make mistakes. That's completely normal — even experienced programmers hit
errors constantly. When you run a cell with an error, Python prints a message telling you (more or less)
what went wrong.

We've planted an error in the next cell on purpose. Run it and see what happens.

```python
print("This line is missing something."
```

You should see a `SyntaxError`, with a message ending in something about "EOF" (end of file) or an
unclosed parenthesis. That's Python's way of saying "you started something and never finished it" — in
this case, a missing closing parenthesis `)`.

You don't need to memorize error terminology to program effectively. If a message looks cryptic, that's
normal too — ask a neighbor or your instructor rather than trying to decode it alone.

**Try it.** Fix the cell above so it runs without an error and prints the intended message.

One more thing worth knowing: the **Run** menu in the toolbar has a **Run All Cells** option, which runs
every code cell in the notebook top to bottom. If a cell errors out, the notebook stops there — so a
single broken cell early in a notebook can block everything after it from running.

### 1.5 The kernel

The **kernel** is the program that actually executes the code in your notebook and produces the results.
In the top right of the window, a small circle shows the kernel's status: empty/hollow means it's idle and
ready, filled-in means it's busy running something.

If your kernel ever gets stuck — a cell runs forever, the notebook feels frozen, or you lose your
connection — try these steps in order:

1. **Kernel** menu → **Interrupt**. This stops whatever's currently running without losing your variables.
2. If that doesn't help: **Kernel** menu → **Restart Kernel**. This clears all your variables, so you'll
   need to re-run your cells from the top.
3. If that still doesn't help: save your work (**File** → **Save Notebook**), then go back to the Hub's
   control panel, stop your server, and start it again. You'll need to re-run your cells once you're back
   in, but your notebook file itself will still be there.
